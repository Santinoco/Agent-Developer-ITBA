## 3. Control de flujo

Las condiciones deciden qué camino tomar. Los ciclos procesan varios elementos sin repetir manualmente el código.


### Antes de ejecutar — tomar una decisión

**Qué observar:** Python ejecuta únicamente la primera condición verdadera.

## 4. Funciones y type hints

Una función concentra una responsabilidad. El nombre, los parámetros y el tipo de retorno forman un contrato legible.


### Antes de ejecutar — encapsular lógica en funciones

**Qué vas a hacer:** separar una estimación reutilizable y una función que presenta mensajes.  
**Qué observar:** `texto: str` describe la entrada y `-> int` describe el retorno.  
**Importante:** esta estimación es didáctica; no reemplaza el contador real del proveedor.


````python
def estimar_tokens(texto: str) -> int:
    # Estimación didáctica: aproximadamente 1 token cada 0.75 palabras.
    palabras = len(texto.split())
    return max(1, round(palabras / 0.75))


def resumir_mensaje(mensaje: dict[str, object], prefijo: str = "Mensaje") -> str:
    contenido = str(mensaje.get("content", ""))
    return f"{prefijo}: {contenido[:45]} | ~{estimar_tokens(contenido)} tokens"


for mensaje in mensajes:
    print(resumir_mensaje(mensaje))
````

````python
assert estimar_tokens("hola mundo") >= 1
assert resumir_mensaje({"content": "Hola"}).startswith("Mensaje:")
print("✓ Checkpoint 4 superado: funciones")
````

## 5. Manejo de errores

El objetivo no es ocultar errores, sino convertir fallas esperables en resultados comprensibles.


### Antes de ejecutar — controlar errores esperables

**Qué vas a hacer:** leer una configuración que puede venir incompleta o contener valores inválidos.  
**Qué observar:** cada `except` responde a una causa concreta y devuelve un valor seguro.  
**Resultado esperado:** primero `0.4`, después dos casos controlados con `0.0`.


````python
def leer_temperature(config: dict[str, object]) -> float:
    try:
        valor = float(config["temperature"])
        if not 0 <= valor <= 2:
            raise ValueError("temperature debe estar entre 0 y 2")
        return valor
    except KeyError:
        print("No se encontró temperature; se usará 0.0")
        return 0.0
    except (TypeError, ValueError) as error:
        print(f"Configuración inválida: {error}")
        return 0.0


print(leer_temperature({"temperature": "0.4"}))
print(leer_temperature({}))
print(leer_temperature({"temperature": 5}))
````

### 5.1 Retry controlado

Este ejemplo simula una operación que falla dos veces y funciona en el tercer intento.


### Antes de ejecutar — reintentar sin crear un loop infinito

**Qué vas a hacer:** simular dos fallas transitorias y detener el ciclo cuando la operación funciona.  
**Qué mirar:** el máximo de intentos está limitado y `break` corta el `for` al obtener una respuesta.  
**Resultado esperado:** dos fallas y un éxito en el tercer intento.


````python
intento_actual = 0


def operacion_inestable() -> str:
    global intento_actual
    intento_actual += 1
    if intento_actual < 3:
        raise ConnectionError("Falla transitoria simulada")
    return "respuesta correcta"


resultado = None
for intento in range(1, 4):
    try:
        resultado = operacion_inestable()
        print(f"Éxito en el intento {intento}: {resultado}")
        break
    except ConnectionError as error:
        print(f"Intento {intento} falló: {error}")

assert resultado == "respuesta correcta"
````

## 7. Proyecto integrador — Analizador de conversaciones

Esta celda combina estructuras, funciones, validaciones, dataclasses y manejo de errores. Ejecutala completa y luego probá agregar un mensaje.


### Antes de ejecutar — integrar lo aprendido

**Qué vas a construir:** una conversación que valida roles, evita textos vacíos, estima tokens y genera un resumen.  
**Cómo recorrer el código:** leé primero los atributos; después `agregar`; finalmente los métodos de cálculo y resumen.  
**Resultado esperado:** tres mensajes impresos y un diccionario con cantidad, tokens y roles.


````python
from dataclasses import dataclass, field


@dataclass
class Conversacion:
    nombre: str
    mensajes: list[Mensaje] = field(default_factory=list)

    def agregar(self, role: str, content: str) -> None:
        if role not in {"user", "assistant", "system"}:
            raise ValueError(f"Role no permitido: {role}")
        if not content.strip():
            raise ValueError("El contenido no puede estar vacío")
        self.mensajes.append(
            Mensaje(role=role, content=content.strip(), tokens=estimar_tokens(content))
        )

    def total_tokens(self) -> int:
        return sum(mensaje.tokens for mensaje in self.mensajes)

    def resumen(self) -> dict[str, object]:
        return {
            "nombre": self.nombre,
            "cantidad_mensajes": len(self.mensajes),
            "total_tokens_estimados": self.total_tokens(),
            "roles": sorted({m.role for m in self.mensajes}),
        }


conversacion = Conversacion("demo-precurso")
conversacion.agregar("user", "Explicame qué es un entorno virtual")
conversacion.agregar("assistant", "Es un espacio aislado para las dependencias del proyecto")
conversacion.agregar("user", "¿Cómo lo activo?")

for item in conversacion.mensajes:
    print(f"[{item.role}] {item.content} (~{item.tokens} tokens)")

print("\nResumen:", conversacion.resumen())
````
